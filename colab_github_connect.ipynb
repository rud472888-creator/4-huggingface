## 2. GitHub 저장소 clone

GitHub token은 입력창에 붙여넣으세요. 화면에 보이지 않습니다.

```python
from getpass import getpass
from pathlib import Path
from urllib.parse import quote
import subprocess

GITHUB_USER = "rud472888-creator"
REPO_NAME = "4-huggingface"
BRANCH_NAME = "colab-work"
BASE_DIR = Path('/content/drive/MyDrive/git')
REPO_DIR = BASE_DIR / REPO_NAME

def run(cmd, cwd=None, mask=None):
    shown = cmd
    if mask:
        shown = [part.replace(mask, '***') for part in cmd]
    print('$', ' '.join(shown))
    subprocess.run(cmd, cwd=cwd, check=True)

BASE_DIR.mkdir(parents=True, exist_ok=True)
token = quote(getpass('GitHub token: ').strip(), safe='')
auth_url = f'https://x-access-token:{token}@github.com/{GITHUB_USER}/{REPO_NAME}.git'
clean_url = f'https://github.com/{GITHUB_USER}/{REPO_NAME}.git'

if REPO_DIR.exists():
    run(['git', 'remote', 'set-url', 'origin', clean_url], cwd=REPO_DIR)
    run(['git', 'fetch', auth_url], cwd=REPO_DIR, mask=token)
else:
    run(['git', 'clone', auth_url, str(REPO_DIR)], cwd=BASE_DIR, mask=token)
    run(['git', 'remote', 'set-url', 'origin', clean_url], cwd=REPO_DIR)

del token, auth_url
run(['git', 'checkout', BRANCH_NAME], cwd=REPO_DIR)
run(['git', 'status', '--short', '--branch'], cwd=REPO_DIR)
```
